```python
import random, time, math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

random.seed(0)
np.random.seed(0)

class Counter:
    def __init__(self):
        self.comparisons = 0
        self.aux = 0
    def tot(self): return self.comparisons + self.aux

# Sorting with counters
def insertion_sort_count(arr):
    a = arr.copy()
    c = Counter()
    n = len(a)
    for i in range(1, n):
        key = a[i]; c.aux += 1
        j = i - 1; c.aux += 1
        while j >= 0:
            c.comparisons += 1
            if a[j] > key:
                a[j + 1] = a[j]; c.aux += 1
                j -= 1; c.aux += 1
            else:
                break
        a[j + 1] = key; c.aux += 1
    return a, c

def merge_sort_count(arr):
    c = Counter()
    def merge(a, l, m, r):
        n1 = m - l + 1; n2 = r - m
        L = a[l:m+1]; R = a[m+1:r+1]; c.aux += n1 + n2
        i=j=0; k=l
        while i < n1 and j < n2:
            c.comparisons += 1
            if L[i] <= R[j]:
                a[k] = L[i]; c.aux += 1; i += 1
            else:
                a[k] = R[j]; c.aux += 1; j += 1
            k += 1
        while i < n1:
            a[k] = L[i]; c.aux += 1; i += 1; k += 1
        while j < n2:
            a[k] = R[j]; c.aux += 1; j += 1; k += 1

    def ms(a, l, r):
        if l < r:
            m = (l + r) // 2
            ms(a, l, m); ms(a, m+1, r); merge(a, l, m, r)

    a = arr.copy()
    if len(a)>0:
        ms(a, 0, len(a)-1)
    return a, c

# QuickSort implemented via Lomuto but allowing chosen pivot index to be moved to end first
def quick_sort_count_lomuto_with_pivot_index(arr, pivot_index_func, count_median3=False):
    c = Counter()
    a = arr.copy()
    def partition(l, h):
        pidx = pivot_index_func(a, l, h, c) if count_median3 else pivot_index_func(a, l, h)
        c.aux += 1
        a[pidx], a[h] = a[h], a[pidx]; c.aux += 3
        pivot = a[h]; c.aux += 1
        i = l - 1; c.aux += 1
        for j in range(l, h):
            c.comparisons += 1
            if a[j] <= pivot:
                i += 1; c.aux += 1
                a[i], a[j] = a[j], a[i]; c.aux += 3
        a[i+1], a[h] = a[h], a[i+1]; c.aux += 3
        return i+1
    def qs(l, h):
        if l < h:
            p = partition(l, h)
            qs(l, p-1); qs(p+1, h)
    qs(0, len(a)-1 if len(a)>0 else 0)
    return a, c

def pivot_first(a,l,h):
    return l

# median-of-three pivot index with counting: we pass counter and increment comparisons/aux inside
def pivot_median3_with_count(a, l, h, c):
    m = (l + h)//2
    x,y,z = a[l], a[m], a[h]
    c.aux += 3
    # Use a small decision tree and count the comparisons used to determine median
    c.comparisons += 1
    if x <= y:
        c.comparisons += 1
        if y <= z:
            return m
        else:
            c.comparisons += 1
            if x <= z:
                return h
            else:
                return l
    else:
        c.comparisons += 1
        if x <= z:
            return l
        else:
            c.comparisons += 1
            if y <= z:
                return h
            else:
                return m

def quick_sort_count_first(arr):
    return quick_sort_count_lomuto_with_pivot_index(arr, pivot_first)

def quick_sort_count_median3(arr):
    return quick_sort_count_lomuto_with_pivot_index(arr, pivot_median3_with_count, count_median3=True)

# ---------- data generation ----------
def make_list(n, kind):
    if kind == "random":
        return [random.randint(0, n) for _ in range(n)]
    elif kind == "ascending":
        return list(range(n))
    elif kind == "descending":
        return list(range(n,0,-1))
    elif kind == "nearly_sorted":
        a = list(range(n))
        swaps = max(1, n//10)
        for _ in range(swaps):
            i = random.randrange(n); j = random.randrange(n)
            a[i], a[j] = a[j], a[i]
        return a
    elif kind == "many_duplicates":
        return [random.choice([1,2,3,4]) for _ in range(n)]
    else:
        raise ValueError(kind)

# ---------- experiment parameters ----------
ns = [10,50,100,200,500,1000,2000,5000]
kinds = ["random","ascending","descending","nearly_sorted","many_duplicates"]
algorithms = [
    ("Insertion", insertion_sort_count),
    ("Merge", merge_sort_count),
    ("QuickFirst", quick_sort_count_first),
    ("QuickMedian3", quick_sort_count_median3)
]
random_trials = 5

results = []
for kind in kinds:
    for n in ns:
        trials = random_trials if kind=="random" else 1
        for name,func in algorithms:
            comps = []; auxs = []; tots = []; times = []
            for t in range(trials):
                a = make_list(n, kind)
                start = time.perf_counter()
                _, c = func(a)
                end = time.perf_counter()
                comps.append(c.comparisons); auxs.append(c.aux); tots.append(c.tot()); times.append(end-start)
            results.append({
                "kind": kind, "n": n, "algorithm": name,
                "comparisons": sum(comps)/len(comps),
                "aux": sum(auxs)/len(auxs),
                "total": sum(tots)/len(tots),
                "time": sum(times)/len(times)
            })

df = pd.DataFrame(results)

# ---------- plotting routine ----------
def plot_counts(df, kind, name, save=False):
    sub = df[(df.kind==kind) & (df.algorithm==name)].sort_values("n")
    n = sub.n.values
    comps = sub.comparisons.values
    aux = sub.aux.values
    tot = sub.total.values
    plt.figure(figsize=(8,5))
    plt.plot(n, comps, marker='o', label='comparisons')
    plt.plot(n, aux, marker='s', label='auxiliary ops')
    plt.plot(n, tot, marker='^', label='total ops')
    plt.xscale('log'); plt.yscale('log')
    plt.xlabel('n (log scale)'); plt.ylabel('operation counts (log scale)')
    plt.title(f'{name} - {kind}')
    plt.legend(); plt.grid(True, which='both', ls='--', linewidth=0.5)
    if save:
        plt.savefig(f"{name}_{kind}_ops.png", dpi=200)
    plt.show()

# Show plots for each algorithm on random and nearly_sorted (adjust if you want more)
for kind in ["random","nearly_sorted"]:
    for name,_ in algorithms:
        plot_counts(df, kind, name)

# ---------- curve fitting ----------
def fit_models(x, y):
    x = np.array(x, dtype=float); y = np.array(y, dtype=float)
    # model: a * n log2(n) + b * n + c
    X1 = np.vstack([x * np.log2(np.where(x>0, x, 1)), x, np.ones_like(x)]).T
    coeffs1, *_ = np.linalg.lstsq(X1, y, rcond=None)
    pred1 = X1 @ coeffs1
    ss_res1 = np.sum((y-pred1)**2); ss_tot = np.sum((y-y.mean())**2)
    r21 = 1 - ss_res1/ss_tot if ss_tot>0 else float('nan')
    # quadratic fit
    X2 = np.vstack([x**2, x, np.ones_like(x)]).T
    coeffs2, *_ = np.linalg.lstsq(X2, y, rcond=None)
    pred2 = X2 @ coeffs2
    ss_res2 = np.sum((y-pred2)**2)
    r22 = 1 - ss_res2/ss_tot if ss_tot>0 else float('nan')
    return coeffs1, r21, coeffs2, r22, pred1, pred2

fit_reports = []
for kind in kinds:
    for name,_ in algorithms:
        sub = df[(df.kind==kind) & (df.algorithm==name)].sort_values("n")
        x = sub.n.values
        comp_coeffs1, comp_r21, comp_coeffs2, comp_r22, comp_pred1, comp_pred2 = fit_models(x, sub.comparisons.values)
        tot_coeffs1, tot_r21, tot_coeffs2, tot_r22, tot_pred1, tot_pred2 = fit_models(x, sub.total.values)
        fit_reports.append({
            "kind": kind, "algorithm": name,
            "comp_nlogn_coeffs": comp_coeffs1.tolist(), "comp_nlogn_r2": comp_r21,
            "comp_quad_coeffs": comp_coeffs2.tolist(), "comp_quad_r2": comp_r22,
            "tot_nlogn_coeffs": tot_coeffs1.tolist(), "tot_nlogn_r2": tot_r21,
            "tot_quad_coeffs": tot_coeffs2.tolist(), "tot_quad_r2": tot_r22
        })

fit_df = pd.DataFrame(fit_reports)

# print fit results for random lists
print("Fit results (random lists):")
print(fit_df[fit_df.kind=='random'].to_string(index=False))

# ---------- compare time vs total ops correlation ----------
print("\nCorrelation (total operations vs measured time) for random lists:")
for name,_ in algorithms:
    sub = df[(df.kind=="random") & (df.algorithm==name)].sort_values("n")
    corr = np.corrcoef(sub.total.values, sub.time.values)[0,1]
    print(f"{name}: corr(total_ops, time) = {corr:.4f}")

# ---------- show main results for random lists ----------
print("\nSample of results (random lists):")
print(df[df.kind=="random"].sort_values(["algorithm","n"]).to_string(index=False))
```
