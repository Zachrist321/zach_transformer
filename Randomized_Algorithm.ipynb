```python
# Randomized Quicksort is a Las Vegas Algorithm
arr = [8, 2, 3, 4, 10, 11, -10]
def randomize_sort(arr):
    if len(arr) < 2:
        return arr
    pivot = np.random.randint(0, len(arr))
    arr[pivot], arr[-1] = arr[-1], arr[pivot]
    left = [i for i in arr[:-1] if i <= arr[-1]]
    right = [i for i in arr[:-1] if i > arr[-1]]
    return randomize_sort(left) + [arr[-1]] + randomize_sort(right)
```
