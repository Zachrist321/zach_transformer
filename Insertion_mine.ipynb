```python
coin_tosses = []
for i in range(100):
    coin_tosses.append(np.random.choice([0, 1]))
```

```python
num_tail = coin_tosses.count(0)
probability_tail = num_tail / len(coin_tosses)
probability_tail
```

```python
# Inserstion Sort
A = [-10, 9, 1, 9, 10, -111]
for i in range(1, len(A)):
    j = i - 1
    while j >= 0:
        if A[j] >= A[j+1]:
            A[j], A[j+1] = A[j+1], A[j]
        j -= 1
```
