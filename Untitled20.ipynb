```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# You can use this cell for your calculations (not graded)
dice = [1, 2, 3, 4, 5, 6]
out_comes = []
for i in range(1000):
    first = np.random.choice(dice)
    second = np.random.choice(dice)
    sum = first + second
    out_comes += sum
bins = np.arange(1.5, 13, 1)
```

```python
fig, ax = plt.subplots()
ax.hist(out_comes)
plt.show()
```
