## GCD

```python
# Returns the greatest common divisor of a and b
def gcd(a, b):
    while b:
        a, b = b, a % b
    return a
```

## EGCD

```python
# Returns (gcd, x, y) such that ax + by = gcd(a, b)
def extended_gcd(a, b):
    if b == 0:
        return (a, 1, 0)
    g, x1, y1 = extended_gcd(b, a % b)
    x, y = y1, x1 - (a // b) * y1
    return g, x, y
```

## Modular Inverse

```python
# Returns the modular inverse of a modulo m, if it exists
def mod_inverse(a, m):
    g, x, _ = extended_gcd(a, m)
    if g != 1:
        raise ValueError("Inverse does not exist")
    return x % m
```

## RSA Encryption

```python
# Generates RSA public and private keys from primes p, q and exponent e
def rsa_keygen(p, q, e):
    n = p * q
    phi = (p - 1) * (q - 1)
    if gcd(e, phi) != 1:
        raise ValueError("e and phi(n) must be coprime")
    d = mod_inverse(e, phi)
    return (e, n), (d, n)
    
# Encrypts a string message using the RSA public key
def rsa_encrypt(msg, pubkey):
    e, n = pubkey
    return [pow(ord(c), e, n) for c in msg]
    
# Decrypts an RSA-encrypted list of integers using the private key
def rsa_decrypt(cipher, privkey):
    d, n = privkey
    return ''.join([chr(pow(c, d, n)) for c in cipher])
```

## Parity Check For Error Detection

```python
# Appends an even parity bit to the binary string data_bits
def add_parity_bit(data_bits):
    parity = sum(int(b) for b in data_bits) % 2
    return data_bits + str(parity)
    
# Checks if the received binary string has correct even parity
def check_parity(received_bits):
    total = sum(int(b) for b in received_bits)
    return total % 2 == 0
```
