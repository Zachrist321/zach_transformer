```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
arr = np.random.choice([0, 1], (10))
```

```python
arr
```

```python
count = 0
for element in arr:
    if element == 1:
        count += 1
```

```python
count / len(arr)
```

```python
x = [1, 2, 3, 4]
nums = [(1 / 8), (3 / 8), (3 / 8), (1 / 8)]
nums = [i * 100 for i in nums]
nums
```

```python
fig, ax = plt.subplots()
ax.hist(x, weights=nums)
plt.show()
```
