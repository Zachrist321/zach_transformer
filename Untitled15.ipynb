```python
def binary_search(arr, word):
    if len(arr) < 2:
        print("It is either the arr you put is small or after dividing, it couldn't find it till the array length is less than 2")
        return None
    mid = len(arr) // 2
    if word == arr[mid]:
        print("we found the thing")
        return None
    elif word < arr[mid]:
        print("we are going to the left")
        binary_search(arr[:mid], word)
    else:
        print("we are going to the right")
        binary_search(arr[mid:], word)
```

```python
import numpy as np
```

```python
arr = np.arange(100)
```

```python
binary_search(arr, 100)
```
