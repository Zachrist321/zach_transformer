```python
arr = ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h', 'i', 'j', 'k', 'l']
def binary_search(arr, letter):
    if len(arr) <= 1:
        return f"finished at the first check"
    m = len(arr) // 2
    if letter > arr[m - 1]:
        print(f"checking greater than {m}")
        binary_search(arr[m:], letter)
    elif letter == arr[m - 1]:
        print(f"guess I found the letter")
        return
    else:
        print(f"checking less than {m}")
        binary_search(arr[:m], letter)
        
```

```python
binary_search(arr, 'b')
```
