```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
random_x = np.random.randint(0, 100, (10000))
ls_x = list(random_x)
```

```python
each_x = set(random_x)
```

```python
fig, ax = plt.subplots()
for each in each_x:
    count_e = ls_x.count(each)
    ax.scatter(each, count_e)
plt.show()
    
```
