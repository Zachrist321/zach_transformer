```python
import numpy as np
A = [9, 9, 8, 7, 7, 9, 1]
b = [8, 7]
lenA = len(A)
lenb = len(b)
output_list = []
```

```python
x = 0
for i in range(lenb - 1, -1, -1):
    intrim = []
    car = 0
    for j in range(lenA - 1, -1, -1):
        point_mul = b[i] * A[j] + car
        car = point_mul // 10
        if j != 0:
            intrim.append(point_mul % 10)
        else:
            intrim.append(point_mul)
    each_loop = ""
    for k in range(len(intrim) - 1, -1, -1):
        each_loop = each_loop + str(intrim[k])
    output_list.append(int(each_loop) * 10**x)
    x += 1
        
```

```python
sum(output_list) == 9987791 * 87
```
