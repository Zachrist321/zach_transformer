```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
X = np.array([(1, 2, 3), (4, 5, 7), (1.5, 8.1, 100)])
```

```python
X
```

```python
mean = np.mean(X, axis=0)
std = np.std(X, axis=0)
X = (X - mean) / std
m, n = X.shape
X
```

```python
Y_train = np.arange(m)
```

```python
w = np.ones(n)
b = np.ones(1)
```

```python
def cost(x, y, w, b):
    yp = X @ w + b
    loss = (yp - y) ** 2
    cost = (np.sum(loss)) / (2 * m)
    return cost
    
```

```python
cost(X, Y_train, w, b)
```
