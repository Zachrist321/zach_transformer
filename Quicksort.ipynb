```python
# I will start with a hypothetical array named A
A = [9, 8]
i = -1
pivot = A[-1]
for j in range(0, len(A) - 1):
    if A[j] <= pivot:
        i += 1
        A[i], A[j] = A[j], A[i]
A[i + 1], A[len(A) - 1] = A[len(A) - 1], A[i + 1]
        
```
