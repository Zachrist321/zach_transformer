```python
def karatsuba(x, y):
    if x < 10 or y < 10:
        return x * y        # The base case
    n = max(len(str(x)), len(str(y)))
    m = n // 2
    power = 10 ** m
    high1 = x // power
    low1 = x % power
    high2 = y // power
    low2 = y % power
    z1 = karatsuba(high1, high2)
    z0 = karatsuba(low1, low2)
    z2 = karatsuba(low1 + high1, low2 + high2)
    return z1 * (10 ** (2 * m)) + (z2 - z0 - z1) * (10 ** m) + z0
```
