### 5.1 Design a Three-Wheel Functional Robot (URDF)

```python
<?xml version="1.0"?>
<robot name="ethiopia_bot">
  <!-- Chassis -->
  <link name="base_link">
    <visual><geometry><box size="0.5 0.3 0.1"/></geometry></visual>
    <collision><geometry><box size="0.5 0.3 0.1"/></geometry></collision>
    <inertial><mass value="5.0"/><inertia ixx="0.1" ixy="0" ixz="0" iyy="0.1" iyz="0" izz="0.1"/></inertial>
  </link>

  <!-- Left/Right Wheels and Caster omitted for brevity but standard in URDF -->

  <!-- Sensors: Camera -->
  <link name="camera_link">
    <visual><geometry><box size="0.05 0.05 0.05"/></geometry></visual>
  </link>
  <joint name="camera_joint" type="fixed">
    <parent link="base_link"/><child link="camera_link"/><origin xyz="0.25 0 0.075"/>
  </joint>

  <!-- Gazebo Plugins for Differential Drive and Sensors would go here -->
  <gazebo>
    <plugin name="differential_drive_controller" filename="libgazebo_ros_diff_drive.so">
      <leftJoint>left_wheel_joint</leftJoint>
      <rightJoint>right_wheel_joint</rightJoint>
      <robotBaseFrame>base_link</robotBaseFrame>
      <commandTopic>cmd_vel</commandTopic>
    </plugin>
  </gazebo>
</robot>
```
