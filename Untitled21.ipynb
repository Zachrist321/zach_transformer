```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
sam = np.arange(1, 1000)
```

```python
sample_mean = []
```

```python
for _ in range(30):
    x = np.random.choice(sam, (20))
    x_mean = np.mean(x)
    sample_mean.append(x_mean)
```

```python
np.mean(sample_mean)
```

```python
fig, ax = plt.subplots()
ax.hist(sample_mean)
plt.show()
```
