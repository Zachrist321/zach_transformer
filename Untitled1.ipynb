```python
arr = np.zeros((6, 6), dtype=object)
```

```python
arr
```

```python
count = 0
for i in range(6):
    for j in range(6):
        arr[i, j] = (i + 1, j + 1)
        if (i + j + 2) == 7:
            count += 1
        if j - i == 1 and (i + j + 2) != 7:
            count += 1
        if i - j == 1 and (i + j + 2) != 7:
            count += 1
```
