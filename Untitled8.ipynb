```python
import numpy as np
```

```python
def pmf(n):
    j = np.array([0, 1])
    output_active = []
    for i in range(n):
        x = np.random.choice(j)
        if x == 1:
            output_active.append(x)
    return len(output_active)
```

```python
output = []
for i in range(100):
    output.append(pmf(100) / 100)

np.sum(output) / 100
```
