# Uninformed Search Algorithm

```python
graph = {
    'A': ['B', 'C', 'D'],
    'B': ['E', 'F'],
    'C': ['G'],
    'D': ['H', 'I'],
    'E': [],
    'F': [],
    'G': [],
    'H': [],
    'I': [],
}
```

```python
from collections import deque
```

```python
# Breadth-First Search (BFS) using a queue
def bf(graph, start):
    visited = set()
    queue = deque([start])
    print("BFS")

    while queue:
        node = queue.popleft()
        print(node, end=" ")
        if node == 'G':
            break

        visited.add(node)
        for neighbor in graph[node]:
            if neighbor not in visited:
                queue.append(neighbor)
            
        
```

```python
bf(graph, 'A')
```

# Depth First

```python
def dfs(graph, start):
    stack = [start]
    visited = set()
    print("DFS Traversal:")

    while stack:
        node = stack.pop()
        print(node, end=' ')

        if node not in visited:
            visited.add(node)

        for neighbor in reversed(graph[node]):
            stack.append(neighbor)
```

```python
dfs(graph, 'A')
```
